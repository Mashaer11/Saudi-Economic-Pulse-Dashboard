```python
import pandas as pd
import numpy as np
from statsmodels.tsa.arima.model import ARIMA
import warnings
warnings.filterwarnings('ignore')

# 1. إنشاء السلسلة الزمنية للمؤشرات السعودية (من يناير 2021 إلى منتصف 2026)
np.random.seed(42)
dates = pd.date_range(start='2021-01-01', end='2026-06-01', freq='MS')

# محاكاة اتجاه النمو الاقتصادي لمبيعات نقاط البيع والتجارة الإلكترونية في السعودية
n_periods = len(dates)
trend_pos = np.linspace(35, 62, n_periods)  # نمو مبيعات نقاط البيع (بالمليار ريال)
trend_ecom = np.linspace(8, 24, n_periods)  # نمو مبيعات التجارة الإلكترونية مدى (بالمليار ريال)

seasonality = 3 * np.sin(np.pi * dates.month / 6)
noise = np.random.normal(0, 1.2, n_periods)

pos_actual = np.round(trend_pos + seasonality + noise, 2)
ecom_actual = np.round(trend_ecom + (seasonality * 0.8) + noise, 2)

# إنشاء DataFrame الأساسي
df = pd.DataFrame({
    'Date': dates,
    'Year': dates.year,
    'Quarter': dates.to_period('Q').astype(str),
    'POS_Sales_Billion_SAR': pos_actual,
    'ECom_Sales_Billion_SAR': ecom_actual,
    'Type': 'Actual'
})

# 2. التنبؤ الإحصائي للأرباع القادمة باستخدام نموذج ARIMA
model_pos = ARIMA(df['POS_Sales_Billion_SAR'], order=(1, 1, 1)).fit()
forecast_pos = np.round(model_pos.forecast(steps=6), 2)

model_ecom = ARIMA(df['ECom_Sales_Billion_SAR'], order=(1, 1, 1)).fit()
forecast_ecom = np.round(model_ecom.forecast(steps=6), 2)

future_dates = pd.date_range(start='2026-07-01', periods=6, freq='MS')
df_forecast = pd.DataFrame({
    'Date': future_dates,
    'Year': future_dates.year,
    'Quarter': future_dates.to_period('Q').astype(str),
    'POS_Sales_Billion_SAR': forecast_pos.values,
    'ECom_Sales_Billion_SAR': forecast_ecom.values,
    'Type': 'Forecast'
})

# دمج البيانات الفعلية مع التوقعات
final_df = pd.concat([df, df_forecast], ignore_index=True)

# 3. حفظ البيانات وتنزيلها كملف CSV جاهز لـ Tableau
final_df.to_csv('saudi_economic_pulse_clean.csv', index=False)
print("✅ تم تجهيز السلسلة الزمنية والتنبؤ بنجاح! الملف جاهز للتحميل: saudi_economic_pulse_clean.csv")
```
